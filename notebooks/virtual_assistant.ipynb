```python
from gtts import gTTS

def textToSpeech(text_to_say, language, path):
  gtts = gTTS(text = text_to_say, lang = language, slow = False)
  gtts.save(path)
```

Testing textToSpeechFunction

```python
from IPython.display import Audio
```

2. Virtual assistant

```python
import speech_recognition as sr
```

```python
# Listen to a command from mic
def listen():
    r = sr.Recognizer()
    r.pause_threshold = 1
    r.adjust_for_ambient_noise(source, duration=1)
    said = ''
    with sr.Microphone() as source:
        audio = r.listen(source)
        try:
            said = r.recognize_google(audio)
            print(said)
        except sr.UnknownValueError:
            speak('Sorry, I could not understand waht you said! Could you repeat please')
        except sr.RequestError:
            speak('Sorry, the service is unavailable')
    return said.lower()
```
